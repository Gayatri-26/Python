#**While loop Assignment**

```python
#1.Reverse number

num = 12345
reverse = 0
while num > 0:
 digit = num % 10
 reverse = reverse * 10 + digit
 num = num  //10
print("Reverse number:", reverse)
```

```python
#2.Check Whether a Number is Palindrome

num = 12321
original = num
reverse = 0
while num > 0:
 digit = num % 10
 reverse = reverse * 10 + digit
 num = num  //10
if original == reverse:
  print("Number is palindeome")
else :
  print("Not palindrome")
```

```python
#3. Count Digits in a Number

num = 23467
count = 0
while num > 0:
 count = count + 1
 num = num // 10
print("Total digits:", count)
```

```python
#4.Find Sum of Digits

num = 123456
sum = 0

while num > 0:
  digit = num % 10
  sum = sum + digit
  num = num //10
print("Total sum:", sum)
```

```python
#5. Check Whether a Number is Armstrong Number

num = 153
original = num
total = 0

while num > 0:
  digit = num % 10
  total = total + digit ** 3
  num = num //10
if original == total:
  print("Number is armstrong")
else :
  print("Not armstrong")
```

```python
#6. Check Whether a Number is Prime
num = 5
i = 2
is_prime = True

while i < num:
    if num % i == 0:
        is_prime = False
        break
    i = i + 1

if is_prime and num > 1:
    print("Prime number")
else:
    print("Not prime number")
```

```python
#7. Find Factorial of a Number

num = 6
factorial = 1
while num > 0:
  factorial = factorial * num
  num = num - 1
print("Factorial:",factorial)
```

```python
#8. Generate Fibonacci Series

n = 10
x = 0
y = 1
count = 0
while count < n:
  print(x)
  z = x + y
  x = y
  y = z
  count = count + 1
```

```python
#9. Find the Largest Digit in a Number

num = 123456
largest = 0
while num > 0:
  digit = num % 10
  if digit > largest:
    largest = digit
  num = num // 10
print("Largest digit:", largest)
```

```python
#10. Print Numbers Divisible by 3 and 5

i = 1

while i <= 100:
  if i % 3 == 0 and i % 5 == 0:
    print("Number is divisible:",i)
  i += 1
```
