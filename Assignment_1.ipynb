```python
#1.Write a Python program to print all natural numbers from 1 to n. - using while loop

n = int(input('Enter the number:'))
i = 1

while i<=n:
  print(i)
  i +=1
```

```python
#2.Write a Python program to print all natural number in reverse from n to 1 using while loop

n = int(input('Enter the number:'))
i = n

while i>= 1:
  print(i)
  i -=1
```

```python
#3.Write a Python program to print all alphabets from a to z. - using while loop

ch = ord('a')

end = ord('z')

while ch <= end:
    print(chr(ch))
    ch += 1
```

```python
#6.Write a Python program to find sum of all natural numbers between 1 to n.

n = int(input("Enter a number: "))

i = 1
sum = 0

while i <= n:
    sum += i
    i += 1

print("Sum =", sum)
```
