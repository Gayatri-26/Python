```python
#3.Write a Python program to print all alphabets from a to z. - using while loop

ch = ord('a')

end = ord('z')

while ch <= end:
    print(chr(ch))
    ch += 1
```

```python
#13.Write a Python program to swap first and last digits of a number.

num = input("Enter a number: ")

if len(num) > 1:
    swapped = num[-1] + num[1:-1] + num[0]
else:
    swapped = num

print("Number after swapping =", swapped)
```

```python
#14.Write a Python program to calculate sum of digits of a number.

num = int(input("Enter a number: "))

num = abs(num)
total = 0

while num > 0:
    digit = num % 10
    total += digit
    num //= 10

print("Sum of digits =", total)
```

```python
#15.Write a Python program to calculate product of digits of a number.

num = int(input("Enter a number: "))

num = abs(num)

if num == 0:
    product = 0
else:
    product = 1
    while num > 0:
        digit = num % 10
        product *= digit
        num //= 10

print("Product of digits =", product)
```
